# Track C: Predict the price of a new listing
Loads the saved model pipeline (HistGradientBoosting + preprocessing V2.1) and predicts the nightly price in EUR for one new listing. No training happens here.

```python
import joblib
import pandas as pd

# preprocessing.py must be in the same folder (the saved pipeline uses it)
pipeline = joblib.load("hostwise_pipeline.joblib")
```

## New listing
Change the values below and run the cell.

```python
# --- Enter the new listing here ---
listing = {
    "city": "paris",
    "neighbourhood_cleansed": "Hôtel-de-Ville",
    "room_type": "Entire home/apt",
    "property_type": "Entire rental unit",
    "accommodates": 4,
    "bedrooms": 2,
    "beds": 2,
    "bathrooms_text": "1 bath",
    "latitude": 48.8566,
    "longitude": 2.3522,
    "minimum_nights": 2,
    "review_scores_rating": 4.8,
    "number_of_reviews": 25,
    "host_is_superhost": "f",
    "calculated_host_listings_count": 1,
    "hosts_time_as_host_years": 3,
}
# ----------------------------------

from IPython.display import display, Markdown

price = max(pipeline.predict(pd.DataFrame([listing]))[0], 0)
display(Markdown(
    f"We predict this **{listing['room_type']}** in **{listing['city'].title()}** "
    f"({listing['accommodates']} guests, {listing['bedrooms']} bedrooms) "
    f"costs about **{price:,.0f} EUR** per night."
))
```
